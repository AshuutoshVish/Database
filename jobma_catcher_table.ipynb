```python
for col in catcher_df.columns:
    if catcher_df[col].dtype == 'object':
        catcher_df[col] = catcher_df[col].replace('', pd.NA)
        if catcher_df[col].isna().sum() > 0:
            catcher_df[col] = catcher_df[col].fillna(catcher_df[col].mode().iloc[0])
    else:
        if catcher_df[col].isna().sum() > 0:
            catcher_df[col] = catcher_df[col].fillna(catcher_df[col].median())
```
