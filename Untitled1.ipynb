```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
```

```python
engine = create_engine("mysql+pymysql://root:Jobma009%40@localhost:3306/jobma")
```

```python
jobma_catcher_df = pd.read_sql("SELECT * FROM jobma.jobma_catcher where jobma_catcher_parent=0;", engine)
jobma_catcher_df.shape
```

```python
jobma_catcher_df.rename(columns={'jobma_catcher_id': 'catcher_id'}, inplace=True)
```

```python
catcher_df  = jobma_catcher_df[['catcher_id',
                                'jobma_catcher_title',
                                'jobma_catcher_otype',
                                'jobma_catcher_status',
                                'jobma_catcher_creation',
                                'jobma_catcher_type',
                                'is_premium',
                                'subscription_status',
                                'sub_user',
                                'per_sub_user',
                                'subscription_type',
                                'created_at',
                                'referral_credit',
                                'company_size',
                                'ai_live_interview',
                               ]].copy()
catcher_df
```

```python
# Group-based fill: use the most common value of company_size within each subscription_status
catcher_df['company_size'] = catcher_df.groupby(['subscription_status', 'sub_user','is_premium'])['company_size']\
    .transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else '1-25'))
```

```python
catcher_df['jobma_catcher_type'] = catcher_df.groupby(
    ['subscription_status', 'company_size', 'sub_user', 'is_premium'])['jobma_catcher_type'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else '0'))
```
