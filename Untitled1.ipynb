```python
import os
import torch
import numpy as np
import pandas as pd
from torch import nn
from dotenv import load_dotenv
from sqlalchemy.engine import URL
from sqlalchemy import create_engine
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.base import BaseEstimator, TransformerMixin
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, FunctionTransformer
```

```python
load_dotenv()
db_url = URL.create(
    drivername="mysql+pymysql",
    username=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    host=os.getenv("DB_HOST"),
    port=int(os.getenv("DB_PORT")),
    database=os.getenv("DB_NAME")
)
engine = create_engine(db_url)
```

```python
def clean_catcher_data(engine):
    df = pd.read_sql("""
        SELECT jobma_catcher_id, is_premium, subscription_status, company_size, jobma_catcher_parent 
        FROM jobma_catcher 
        WHERE jobma_verified IN (1, '1')
    """, engine)
    df.replace("", np.nan, inplace=True)
    df['is_premium'] = pd.to_numeric(df['is_premium'], errors='coerce').fillna(0).astype(int)
    df['subscription_status'] = pd.to_numeric(df['subscription_status'], errors='coerce').replace({2: 0}).fillna(0).astype(int)
    sub_accounts = df['jobma_catcher_id'].map(df['jobma_catcher_parent'].value_counts()).fillna(0)
    df['sub_user'] = sub_accounts.astype(int)
    df['company_size'] = df['company_size'].fillna(df['company_size'].mode()[0])
    return df

def clean_wallet_data(engine):
    df = pd.read_sql("SELECT catcher_id AS jobma_catcher_id, wallet_amount, is_unlimited FROM wallet", engine)
    df['wallet_amount'] = pd.to_numeric(df['wallet_amount'], errors='coerce').fillna(0)
    df['is_unlimited'] = pd.to_numeric(df['is_unlimited'], errors='coerce').fillna(0).astype(int)
    return df

def clean_subscription_data(engine):
    df = pd.read_sql("SELECT catcher_id AS jobma_catcher_id, subscription_amount, currency FROM subscription_history", engine)
    df['subscription_amount'] = pd.to_numeric(df['subscription_amount'], errors='coerce').fillna(0)
    df['subscription_amount'] = df.apply(lambda x: round(x['subscription_amount'], 1) if x['currency'] == 0 else round(x['subscription_amount'] / 85, 1), axis=1)
    return df.groupby('jobma_catcher_id').agg(subscription_sum=('subscription_amount', 'sum'), subscription_count=('jobma_catcher_id', 'count')).reset_index()

def clean_login_data(engine):
    df = pd.read_sql("SELECT jobma_user_id AS jobma_catcher_id, jobma_last_login FROM jobma_login", engine)
    df['jobma_last_login'] = pd.to_datetime(df['jobma_last_login'], errors='coerce')
    today = pd.Timestamp.today().normalize()
    df['since_last_login'] = (today - df['jobma_last_login']).dt.days
    df = df.groupby('jobma_catcher_id').agg(since_last_login=('since_last_login', 'min')).reset_index()
    bins = [0, 30, 90, 180, 365, float('inf')]
    labels = [0, 1, 2, 3, 4]
    df['since_last_login'] = pd.cut(df['since_last_login'].fillna(float('inf')), bins=bins, labels=labels, right=False)
    return df

def clean_generic_count(engine, query, id_col, new_col):
    df = pd.read_sql(query, engine)
    return df.groupby(id_col).size().reset_index(name=new_col)

def merge_catcher_related_data(engine):
    catcher_df = clean_catcher_data(engine)
    dfs = [
        clean_wallet_data(engine),
        clean_subscription_data(engine),
        clean_generic_count(engine, "SELECT jobma_catcher_id FROM jobma_pitcher_invitations", 'jobma_catcher_id', 'invitations'),
        clean_generic_count(engine, "SELECT catcher_id AS jobma_catcher_id FROM job_assessment_kit", 'jobma_catcher_id', 'pre_recorded_kit_counts'),
        clean_generic_count(engine, "SELECT jobma_catcher_id FROM jobma_employer_job_posting", 'jobma_catcher_id', 'jobs_posted'),
        clean_generic_count(engine, "SELECT jobma_catcher_id FROM jobma_interviews", 'jobma_catcher_id', 'pre_rec_interview_taken'),
        clean_login_data(engine)
    ]
    for df in dfs:
        catcher_df = catcher_df.merge(df, on='jobma_catcher_id', how='left')
    return catcher_df
```

```python
def working_on_merged_df(df):
    cols = ['invitations', 'pre_recorded_kit_counts', 'jobs_posted', 'pre_rec_interview_taken']
    df[cols] = df[cols].fillna(0).astype(int)
    child_sums = df[df['jobma_catcher_parent'] != 0].groupby('jobma_catcher_parent')[cols].sum()
    for col in cols:
        df.loc[df['jobma_catcher_parent'] == 0, col] += df.loc[df['jobma_catcher_parent'] == 0, 'jobma_catcher_id'].map(child_sums[col]).fillna(0)

    min_login = df[df['jobma_catcher_parent'] != 0].groupby('jobma_catcher_parent')['since_last_login'].min().reset_index().rename(columns={'since_last_login': 'min_login'})
    df = df.merge(min_login, how='left', left_on='jobma_catcher_parent', right_on='jobma_catcher_parent')
    df['since_last_login'] = np.where(df['jobma_catcher_parent'] != 0, df['min_login'], df['since_last_login'])
    df = df[df['jobma_catcher_parent'] == 0].drop(columns=['jobma_catcher_parent', 'min_login'])
    df['wallet_amount'] = df['wallet_amount'].fillna(0)
    df['subscription_sum'] = df['subscription_sum'].fillna(0)
    df['subscription_count'] = df['subscription_count'].fillna(0).astype(int)
    df['is_unlimited'] = df['is_unlimited'].replace('', 1).fillna(0).astype(int)
    df['since_last_login'] = df['since_last_login'].fillna(4).astype(int)
    return df
```

```python
class DataCleaner(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None): return self
    def transform(self, X):
        X = X.copy()
        if 'company_size' in X.columns:
            X['company_size'] = X['company_size'].str.strip().str.lower()
        return X.replace(['None', 'null', 'NaN', ''], np.nan)

def build_preprocessing_pipeline():
    binary_columns = ['is_premium', 'subscription_status', 'is_unlimited']
    ordinal_columns = ['company_size']
    normal_columns = ['subscription_count', 'invitations', 'pre_recorded_kit_counts',
                      'jobs_posted', 'pre_rec_interview_taken']
    log_columns = ['wallet_amount', 'subscription_sum']
    categorical_columns = ['since_last_login']
    company_size_order = ['1-25', '26-100', '101-500', '500-1000', 'more than 1000']

    binary_transformer = Pipeline([('imputer', SimpleImputer(strategy='most_frequent'))])

    ordinal_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('ordinal', OrdinalEncoder(categories=[company_size_order]))
    ])

    normal_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='mean')),
        ('scaler', StandardScaler())
    ])

    log_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='mean')),
        ('log', FunctionTransformer(np.log1p, validate=False)),
        ('scaler', StandardScaler())
    ])

    categorical_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('scaler', StandardScaler())
    ])

    column_transformer = ColumnTransformer([
        ('binary', binary_transformer, binary_columns),
        ('ordinal', ordinal_transformer, ordinal_columns),
        ('normal', normal_transformer, normal_columns),
        ('log', log_transformer, log_columns),
        ('categorical', categorical_transformer, categorical_columns)
    ])

    return Pipeline([
        ('cleaner', DataCleaner()),
        ('preprocessor', column_transformer)
    ])
```

```python
class CustomDataset(Dataset):
    def __init__(self, data):
        self.data = torch.tensor(data, dtype=torch.float32)
    def __len__(self):
        return len(self.data)
    def __getitem__(self, idx):
        return self.data[idx]

BATCH_SIZE = 16
dataset = CustomDataset(X_tensor)
dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
```

```python
class AutoEncoder(nn.Module):
    def __init__(self, input_shape):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_shape, 256),
            nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(128, 64), nn.ReLU(),
            nn.Linear(64, 32)
        )
        self.decoder = nn.Sequential(
            nn.Linear(32, 64), nn.ReLU(),
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 256), nn.ReLU(),
            nn.Linear(256, input_shape)
        )
    def forward(self, X):
        encoded = self.encoder(X)
        decoded = self.decoder(encoded)
        return encoded, decoded

input_dim = X_tensor.shape[1]
autoencoder = AutoEncoder(input_dim)
optimizer = torch.optim.Adam(autoencoder.parameters(), lr=0.001)
criterion = nn.MSELoss()

EPOCHS = 50
for epoch in range(EPOCHS):
    running_loss = 0.0
    for batch in dataloader:
        optimizer.zero_grad()
        encoded, decoded = autoencoder(batch)
        loss = criterion(decoded, batch)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * batch.size(0)
    
    avg_loss = running_loss / len(dataloader.dataset)
    print(f"Epoch [{epoch+1}/{EPOCHS}], Loss: {avg_loss:.4f}")
```
