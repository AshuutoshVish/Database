```python
kits_pre_ai_outer_df = pd.merge(job_assessment_kit_df, ai_video_kit_df,
                           on='catcher_id',
                           how='outer')
```

```python
#job_assessment_kit - 
#Kits created by employers for Pre-recorded interview


kits_pre_ai_outer_df['kits_count'] = kits_pre_ai_outer_df['catcher_id'].map(kits_pre_ai_outer_df['catcher_id'].value_counts())
kits_pre_ai_outer_df=kits_pre_ai_outer_df[['catcher_id','kits_count']]
```
