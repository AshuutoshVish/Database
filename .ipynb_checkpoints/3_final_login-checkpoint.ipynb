```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine("mysql+pymysql://root:Jobma%40123@localhost:3306/jobma")
df = pd.read_sql("SELECT * FROM jobma_login", con=engine)
catcher_df = pd.read_sql("SELECT * FROM jobma_catcher where jobma_catcher_parent=0", con=engine)
```

```python
df.head(2)
```

```python
df=df[['jobma_user_name', 'jobma_created_by', 'jobma_login_id', 'jobma_last_login', 'created_at', 'jobma_login_status']]
```

```python
df['number_of_logins'] =df['jobma_user_name'].map(df['jobma_user_name'].value_counts())
```

```python
# Calculating days
df['jobma_last_login'] = pd.to_datetime(df['jobma_last_login'], errors='coerce')
df['created_at'] = pd.to_datetime(df['created_at'], errors='coerce')

df['days_since_login'] = (pd.Timestamp('today') - df['jobma_last_login']).dt.days
df['days_since_created'] = (pd.Timestamp('today') - df['created_at']).dt.days
```

```python
new_df = catcher_df.merge(df, on='jobma_user_name', how='left')
```
