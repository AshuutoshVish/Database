```python
import pandas as pd
```

```python
catcher_df=pd.read_csv("final_catcher_df.csv")
login_df=pd.read_csv("final_login_df.csv")

# wallet_df=pd.read_csv("final_wallet_df.csv")
wallet_df=pd.read_csv("new_final_wallet_df.csv")
subscription_df=pd.read_csv("final_subscrption_df.csv")

#both Pre_recorded and Ai kits
ai_video_kits_df=pd.read_csv("ai_video_kits_df.csv")
pre_recorded_kits_df=pd.read_csv("pre_recorded_kits_df.csv")

job_posting_df=pd.read_csv("job_posting_df.csv")#
invitation_df=pd.read_csv("job_invitations_df.csv")
pitchers_df=pd.read_csv("pitchers_applied_df.csv") #Number of pitcher client introduce to us

# intervies done #Interview given by candididate
pre_rec_interviews_df=pd.read_csv("pre_rec_interviews_df.csv")
live_interviews_df=pd.read_csv("live_interviews_df.csv") 
ai_interviews_df=pd.read_csv("ai_interviews_df.csv")
```

```python
catcher_login = pd.merge(catcher_df, login_df, 
                         on='catcher_id', 
                         how='left')
catcher_login.shape
```

```python
catcher_login_wallet = pd.merge(catcher_login, wallet_df,
                                on='catcher_id', 
                                how='left')
catcher_login_wallet.shape
```

```python
catcher_login_wallet_sub = pd.merge(catcher_login_wallet, subscription_df, 
                                    on='catcher_id', 
                                    how='left')
catcher_login_wallet_sub.shape
```

```python
catcher_login_wallet_sub_ai = pd.merge(catcher_login_wallet_sub, ai_video_kits_df,
                                    on='catcher_id', 
                                    how='left')
catcher_login_wallet_sub_ai.shape
```

```python
catcher_login_wallet_sub_ai_pre = pd.merge(catcher_login_wallet_sub_ai, pre_recorded_kits_df,
                                    on='catcher_id', 
                                    how='left')
catcher_login_wallet_sub_ai_pre.shape
```

```python
catcher_login_wallet_sub_ai_pre_jobposting = pd.merge(catcher_login_wallet_sub_ai_pre, job_posting_df,
                                                    on='catcher_id', 
                                                    how='left')
catcher_login_wallet_sub_ai_pre_jobposting.shape
```

```python
catcher_login_wallet_sub_ai_pre_jobposting_invite = pd.merge(catcher_login_wallet_sub_ai_pre_jobposting, invitation_df, 
                                                           on='catcher_id', 
                                                           how='left')
catcher_login_wallet_sub_ai_pre_jobposting_invite.shape
```

```python
catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher = pd.merge(catcher_login_wallet_sub_ai_pre_jobposting_invite, pitchers_df, 
                                                                   on='catcher_id',
                                                                   how='left')
catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher.shape
```

```python
catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec = pd.merge(catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher, pre_rec_interviews_df, 
                                                                                 on='catcher_id',
                                                                                 how='left')
catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec.shape
```

```python
catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec_live = pd.merge(catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec, live_interviews_df, 
                                                                                 on='catcher_id',
                                                                                 how='left')
catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec_live.shape
```

```python
catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec_live_ai_df = pd.merge(catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec_live, ai_interviews_df, 
                                                                                 on='catcher_id',
                                                                                 how='left')

catcher_login_wallet_sub_ai_pre_jobposting_invite_pitcher_prerec_live_ai_df.shape
```
