```python
catcher_login = pd.merge(catcher_df, login_df, 
                         on='catcher_id', 
                         how='inner')

catcher_login.shape
```

```python
# catcher_login.isna().sum()
```

```python
catcher_login_wallet = pd.merge(catcher_login, wallet_df, 
                                on='catcher_id', 
                                how='inner')

catcher_login_wallet.shape
```

```python
# catcher_login_wallet.isna().sum()
```

```python
catcher_login_wallet_sub = pd.merge(catcher_login_wallet, subscription_df, 
                                    on='catcher_id', 
                                    how='inner')

catcher_login_wallet_sub.shape
```

```python
catcher_login_wallet_sub.duplicated().sum()
```

```python
# catcher_login_wallet_sub.isna().sum()
```

```python
catcher_login_wallet_sub_kits = pd.merge(catcher_login_wallet_sub, kits_df,
                                    on='catcher_id', 
                                    how='left')

catcher_login_wallet_sub_kits.shape
```

```python
catcher_login_wallet_sub_kits.duplicated().sum()
```

```python
# catcher_login_wallet_sub_kits.isna().sum()
```

```python
catcher_login_wallet_sub_kits_jobposting = pd.merge(catcher_login_wallet_sub_kits, job_posting_df,
                                                    on='catcher_id', 
                                                    how='left')

catcher_login_wallet_sub_kits_jobposting.shape
```

```python
catcher_login_wallet_sub_kits_jobposting_invite = pd.merge(catcher_login_wallet_sub_kits_jobposting, invitation_df, 
                                                           on='catcher_id', 
                                                           how='left')

catcher_login_wallet_sub_kits_jobposting_invite.shape
```

```python
catcher_login_wallet_sub_kits_jobposting_invite_pitcher = pd.merge(catcher_login_wallet_sub_kits_jobposting_invite, pitchers_df, 
                                                                   on='catcher_id',
                                                                   how='left')

catcher_login_wallet_sub_kits_jobposting_invite_pitcher.shape
```

```python
catcher_login_wallet_sub_kits_jobposting_invite_pitcher_interviews_df = pd.merge(catcher_login_wallet_sub_kits_jobposting_invite_pitcher, interviews_def_online_ai_df, 
                                                                                 on='catcher_id',
                                                                                 how='left')

catcher_login_wallet_sub_kits_jobposting_invite_pitcher_interviews_df.shape
```
