```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("development.csv")
```

```python
def plot_value_distribution(df, column_name, title=None, xlabel=None, ylabel=None, palette='viridis'):
    status_counts = df[column_name].value_counts().sort_index()
    plt.figure(figsize=(6, 4))
    sns.barplot(x=status_counts.index, y=status_counts.values)
    plt.title(title or f'Distribution of {column_name}' )
    plt.xlabel(xlabel or column_name)
    plt.ylabel(ylabel or 'Count')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()    
    plt.show()
```

```python
def plot_histogram(df, column_name, bins=10, title=None, xlabel=None, ylabel=None, kde=False, color='blue'):
    plt.figure(figsize=(6, 4))
    sns.histplot(df[column_name], bins=bins, kde=kde, color=color)
    plt.title(title or f'Distribution of {column_name}')
    plt.xlabel(xlabel or column_name)
    plt.ylabel(ylabel or 'Frequency')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()

    # Show the plot
    plt.show()
```
