```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("pre_final_database.csv")
df.drop(columns=['live_interview_credit','pre_recorded_credit','ai_live_interview_credit','referral_credit'],inplace=True)
df['sub_user'] = df['sub_user'].apply(lambda x: 200 if x > 200 else x)
```

```python
ordinal_columns = ['catcher_id','sub_user','days_since_creation','credit_amount','wallet_amount','subscription_amount','credit_given','currency_val','times_of_subscription','kits_count','position_created','number_of_invitations','pitchers_counts','def_interview_counts','interview_counts',]
new_df= df[ordinal_columns].copy()
new_df.shape
```

```python
def plot_histogram(data, columns, kde=True, bins=30, figsize=(6, 3)):
    for col in columns:
        plt.figure(figsize=figsize)
        sns.histplot(data=df, x=col, kde=kde, bins=bins)
        plt.title(f"Distribution of '{col}'")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()
```

```python
def categorize_credit_amount(val):
    if val == 0.0:
        return 'no_credit'
    elif val <= 100:
        return 'low_credit'
    elif val <= 1000:
        return 'medium_credit'
    else:
        return 'extreme high'
        
df['credit_amount_cat'] = df['credit_amount'].apply(categorize_credit_amount)
```

```python
def categorize_sub_user(val):
    if val == 0:
        return 'none'
    elif val == 1:
        return 'very_low'
    elif 2 <= val <= 5:
        return 'low'
    elif 6 <= val <= 20:
        return 'medium'
    elif 21 <= val <= 99:
        return 'high'
    elif 100 <= val <= 199:
        return 'very_high'
    else:
        return 'extreme'
df['sub_user_cat'] = df['sub_user'].apply(categorize_sub_user)
```

```python
plot_histogram(data=df, columns=["credit_amount", "credit_amount_cat"], bins=10)
```

```python
plot_histogram(data=df, columns=["sub_user", "sub_user_cat"], bins=20)
```
