```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sqlalchemy import create_engine
engine = create_engine("mysql+pymysql://root:Jobma009%40@localhost:3306/jobma")
```

```python
df = pd.read_sql("SELECT * FROM wallet", engine)
```

```python
df = df.groupby('catcher_id').agg(wallet_sum=('wallet_amount', 'sum'),
                                  subscription_type=('subscription_type', 'first'),
                                  plan_type=('plan_type', 'max'),
                                  is_unlimited=('is_unlimited', 'max')
                                 ).reset_index()
```
