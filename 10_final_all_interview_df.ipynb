```python
import pandas as pd
from sqlalchemy import create_engine
engine = create_engine("mysql+pymysql://root:Jobma009%40@localhost:3306/jobma")
```

```python
#candidate interview response data
pre_rec_interviews_df = pd.read_sql("SELECT * FROM jobma_interviews", engine) #just for pre-recorded type
live_interviews_df = pd.read_sql("SELECT * FROM jobma_interviews_online", engine)# live interview type
ai_interviews_df = pd.read_sql("SELECT * FROM jobma_ai_interviews", engine)
```

```python
pre_rec_interviews_df.rename(columns={"jobma_catcher_id":"catcher_id"}, inplace=True)
live_interviews_df.rename(columns={"jobma_catcher_id":"catcher_id"}, inplace=True)
ai_interviews_df.rename(columns={"jobma_catcher_id":"catcher_id"}, inplace=True)
```

```python
pre_rec_interviews_df = pre_rec_interviews_df.groupby('catcher_id').size().reset_index(name='pre_rec_interviews_counts')
```

```python
pre_rec_interviews_df
```

```python
live_interviews_df = live_interviews_df.groupby('catcher_id').size().reset_index(name='live_interviews_counts')
```

```python
live_interviews_df
```

```python
ai_interviews_df = ai_interviews_df.groupby('catcher_id').size().reset_index(name='ai_interviews_counts')
```

```python
ai_interviews_df
```

```python
pre_rec_interviews_df.to_csv("pre_rec_interviews_df.csv", index=False)
live_interviews_df.to_csv("live_interviews_df.csv", index=False)
ai_interviews_df.to_csv("ai_interviews_df.csv", index=False)
```
